### Застосування функції, яка визначає тип даних

```python
df = pd.DataFrame({
    "A": [1, "2", 3.0, "4", True],
    "B": [5, "6", 7.0, "8", False]
})
```

```python
df
```

```python
def dtype(x):
    if isinstance(x, int):    # Якщо значення x є цілим числом (наприклад, 5 або -10)
        return "int"          # тоді повертаємо текстовий рядок "int"
    elif isinstance(x, float):    # Інакше, якщо x є числом з крапкою (наприклад, 3.14)
        return "float"            # повертаємо рядок "float"
    elif isinstance(x, str):      # Інакше, якщо x є текстовим рядком (наприклад, "Привіт")
        return "str"              # повертаємо рядок "str"
    elif isinstance(x, bool):     # Інакше, якщо x є логічним значенням (True або False)
        return "bool"             # повертаємо рядок "bool"
    else:                         # Якщо тип не збігається з жодним із перелічених вище
        return "unknown"          # повертаємо рядок "unknown" (невідомо)
        
```

```python
df['A'].apply(dtype)   # Застосувати функцію dtype до кожного елемента у стовпці 'A'
```

### Застосування функції, яка обчислює відсоток від загальної суми

```python
df = pd.DataFrame({
    "A":[10, 20, 30, 40, 50],
    "B":[60, 70, 80, 90, 100],
    "C":[70, 10, 20, 40, 90]
})
```

```python
def percent(x, total):  # Оголошуємо функцію з назвою percent, яка приймає два аргументи: саме число (x) та загальну суму (total)
    return(x/total) * 100    # Ділимо число на загальну суму (отримуємо частку) та множимо на 100, щоб перевести у відсотки
```

```python
df['A'].sum()*100
```

```python
df['A'].apply(percent, total=df['A'].sum())
```

```python
 # перевіряємо чи правильно порахувало суму, має вивести 100
df['A'].apply(percent, total= df['A'].sum()).sum()
```

```python
df.apply(percent, axis=1, total = df.sum())  # застосування до всіх колонок
```

```python
df.sum()  #Сума всіх колонок
```

```python
df.apply(percent, axis=1, total = df.sum()).sum()  # Перевірити себе. Щоб по всіх колонках було 100
```
