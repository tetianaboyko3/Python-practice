## 🔍**Валідація даних**
У нас є колонка `Total Volume`, яка має бути рівною сумі всіх проданих авокадо, в сумочках і поштучно. Перевіримо це.

````python
df[['Small Number', 'Large Number', 'XLarge Number', 'Total Bags']].sum(axis=1)
````

````python
df['Total Volume'].equals(df[['Small Number','Large Number', 'XLarge Number', 'Total Bags']].sum(axis=1))
````

🕵️‍♂️ Команда  df['Total Volume'].equals(...) перевіряє, чи є колонка Total Volume абсолютно ідентичною до порахованої вами суми (з точністю до типу даних та кожного знака після коми).

💡 Не всюди умова виконується- вивело False.

Подивимось, в скількох даних не виконується.
Для початку запишемо обчислення в окрему колонку.

````python
df['Check Sum'] = df[['Small Number', 'Large Number', 'XLarge Number', 'Total Bags']].sum(axis=1)
````

Тепер маємо порівняти наші значення в колонках `Total Volume` і `Check Sum`. Зробимо це спеціальним методом з бібліотеки `numpy`.

````python
df['Check Passed']= np.isclose(df['Total Volume'], df['Check Sum'])
````

Функція **np.isclose** з бібліотеки NumPy — це головний інструмент для порівняння чисел із плаваючою комою (тип float).Вона перевіряє, чи є два числа (або два масиви чисел) приблизно рівними з урахуванням крихітної комп'ютерної похибки.

````python
np.isclose(12.5, 12.500000001), 12.5 == 12.500000001
````

📊 Подивимось чи розбіжність між ообчисленою колонкою і оригінальною - велика. Побудуємо графік


````python
filtered = df[df['Check Passed']==False]  # Фільтруємо датафрейм: залишаємо лише ті рядки, де перевірка НЕ пройшла
(
    (filtered['Total Volume'] - filtered['Check Sum'])/filtered['Check Sum']
).hist();   # Обчислюємо відносне відхилення (похибку) між двома колонками та будуємо гістограму
````

###  🧩🧹 Агрегація і фільтрація даних

**1. Авокадо якого типу продаються більше? 🥑**

````python
# 1. Групуємо датафрейм за колонкою 'type' (розділяємо дані на органічні та звичайні авокадо)
# 2. Через .agg() створюємо результуючу колонку 'avg_vol'
# 3. Рахуємо середнє значення ('mean') для існуючої колонки 'Total Volume'

df_grouped= df.groupby('type').agg(avg_vol=('Total Volume Errors','mean'))
````

````python
df_grouped
````

````python
# 1. Беремо значення 'avg_vol' для кожного типу і ділимо на СУМУ всіх середніх об'ємів у таблиці
# 2. Множимо на 100, щоб перевести отриману частку у відсотки
# 3. Результат записуємо в нову колонку з назвою 'percentage'

df_grouped['percentage'] = (df_grouped['avg_vol']/df_grouped['avg_vol'].sum())*100
````

````python
df_grouped
````

````python
df_grouped.round(2)  # Округляємо всі числові значення в df_grouped до 2 знаків після коми
````

**3. В якому регіоні найдешевше і найдорожче звичайне авокадо?🗺️💸💎**

Для цього маємо спочатку відфільтрувати дані, а потім загрегувати і відсортувати.

````python
df_conv = df[df['type']=='conventional']   # створює новий відфільтрований датафрейм df_conv, який містить записи виключно про звичайне авокадо
````

````python
# 1. Групуємо відфільтровані дані про звичайне авокадо за колонкою 'region'
# 2. Беремо лише колонку 'AveragePrice' (середня ціна)
# 3. Рахуємо середнє значення (.mean()) для кожного регіону
# 4. Сортуємо отримані ціни за зростанням (.sort_values())

df_conv_by_region= df_conv.groupby('region').AveragePrice.mean().sort_values()
````

````python
df_conv_by_region
````

Знайдемо максимальне і мінімальне значення.

````python
df_conv_by_region.min(),df_conv_by_region.max()
````

````python
# Знаходимо назви регіонів з мінімальним і максимальним значеннями

display(
    df_conv_by_region[df_conv_by_region==df_conv_by_region.min()],
    df_conv_by_region[df_conv_by_region==df_conv_by_region.max()]
)
````

#**📈🥑 Візуалізація даних**

Візуалізуємо і порівняємо динаміку цін на авокадо в цілому, а також за типами.

````python
# Оскільки колонка Date має тип даних object (текст), то
# Перетворимо стовпець "Date" на об'єкт datetime і відсортуємо DataFrame за датою

df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')
df = df.sort_values(by='Date')
````

````python
df['Date']   # дивимося на тип даних в колонці Date- уже стало datetime
````

Побудова візуалізацій. Для початку імпортуємо дві бібліотеки які нам потрібні.

````python
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# Побудуємо найпростіший графік
plt.figure(figsize=(12,6))
plt.plot(df.Date, df.AveragePrice);
````

Так сталось, бо у нас дані по країнам. Відповідно дані "скачуть", бо обʼєми продажів у нас в різних країнах і для різних типів авокадо різні. "Приблизимо трохи дані"

Що ж робити? Ну, наприклад, агрегувати дані.

````python
plt.figure(figsize=(12,6))
plt.plot(df.groupby('Date').AveragePrice.mean());
````

````python
df.groupby('Date').AveragePrice.mean()
````

````python
plt.figure(figsize=(12, 6))
sns.lineplot(x='Date',     # по осі X відкладаємо дати
             y='AveragePrice',  # по осі Y відкладаємо середню ціну
             data=df,           # вказуємо наш початковий датафрейм як джерело даних
             alpha=0.3)         # робить лінії на 70% прозорими, щоб зменшити візуальне навантаження від великої кількості точок
plt.title('Price Trend');
````

````python
# Виведемо динаміку по типах авокадо
plt.figure(figsize=(12, 6))     # Створюємо полотно для графіка: задаємо ширину 12 та висоту 6 дюймів
sns.lineplot(x='Date', y='AveragePrice', hue='type', data=df, alpha=0.3)   #hue='type' — розділяємо дані за кольором залежно від типу авокадо (organic / conventional)
plt.title('Price Trend by Type');   # Додаємо заголовок до нашого графіка
````

💡 **Висновок:**

* **Органічні авокадо:** аналіз довгострокової динаміки цін показує, що вони є **дорожчими**.
* **Звичайні авокадо:** аналіз змін цін за тривалий період свідчить, що вони є **дешевшими**.
