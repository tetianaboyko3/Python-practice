## GroupBy

```python
# Створення словника, де для кожної назви стовпця прописано список значень, і перетворити цей словник на таблицю Pandas
df = pd.DataFrame({
    "A": [7, 7, 5, 1, 1, 5, 1, 9, 1, 8, 5, 7, 3, 4, 3, 7, 7, 6, 9, 7],
    "B": [1, 2, 3, 1, 1, 2, 4, 1, 3, 4, 4, 3, 3, 4, 1, 3, 4, 3, 4, 1],
    "C": ['green', 'green', 'red', 'blue', 'blue', 'green', 'red', 'red',
       'blue', 'green', 'red', 'green', 'blue', 'red', 'blue', 'blue',
       'blue', 'green', 'green', 'red']
})
```

```python
df.head()
```

```python
grouped= df.groupby('A')   # Групування за стовпцем "A"
```

```python
# Показуємо згруповані дані
grouped
```

```python
grouped.size()  # повертає кількість елементів у кожній групі
```

```python
# Перевіряємо кількість рядків у стовпці 'A', де значення дорівнює 1
df[df.A==1].shape[0]
```

```python
# Перевіряємо кількість рядків у стовпці 'A', де значення дорівнює 7
df[df.A==7].shape[0]
```

```python
grouped.mean()   # помилка: не вказали колонки
```

```python
df[df.A==1].B  # Покажи  всі числа зі стовпця B, які стоять навпроти одиниць у стовпці A
```

```python
df[df.A==1].B.mean()   # обчислює середнє арифметичне значення стовпця 'B' лише для тих рядків, де у стовпці 'A' стоїть одиниця
#1+1+4+3= 9/4= 2.25
```

```python
# Виправляємо помилку. Рахуємо середнє
grouped.mean(numeric_only=True) 
```

```python
grouped['B'].mean()  # Обчислення середнього значення стовпця 'B' для кожної групи зі стовпця 'A'
```

```python
df.groupby('A').min()
```

```python
df.groupby('A').describe()  # Згрупуй таблицю за стовпцем 'A' і виведи повний набір описової статистики для всіх числових стовпців
```

```python
df.groupby('A').agg('max')   # Згрупуй дані за стовпцем 'A' і обчисли максимальне значення для кожного іншого стовпця
```

```python
df.groupby('A')['B'].agg(['size', 'max', 'min','mean', 'std'])
```

```python
def min_minus_one(values):
    return values.min()-1
```

```python
df.groupby('A')['B'].agg(min_minus_one)   # Для кожної групи значень 'A' знайди мінімум у стовпці 'B' і відніми від нього 1
```

```python
# Як перевірити себе?
df[df.A==1].B
```

df.groupby('A'): Ми ділимо таблицю на групи за унікальними числами в стовпці 'A'.

.agg({ ... }): Ми передаємо словник, де вказуємо «завдання» для кожного стовпця окремо:

'B': max — для стовпця 'B' знайти лише максимальне значення.

'C': [pd.Series.mode] — для стовпця 'C' (текст) знайти моду

```python
df.groupby('A').agg({'B': max, 'C': [pd.Series.mode]})
```

## Функція pivot()

```python
df = pd.DataFrame({'foo': ['one', 'one', 'one', 'two', 'two', 'two'],
                    'bar': ['A', 'B', 'C', 'A', 'B', 'C'],
                    'baz': [1, 2, 3, 4, 5, 6],
                    'zoo': ['x', 'y', 'z', 'q', 'w', 't']})
```

```python
df
```

```python
# Створюємо новий датафрейм, групований за стовпцем "A"
pivoted= df.pivot(index='foo', columns='bar', values = 'baz')
```

```python
pivoted
```

```python
df.pivot(index='foo', columns= 'bar') #якщо не вкажемо values
```

## Функція pivot_table()

```python
df = pd.DataFrame({"A": ["foo", "foo", "foo", "foo", "foo",
                          "bar", "bar", "bar", "bar"],
                    "B": ["one", "one", "one", "two", "two",
                          "one", "one", "two", "two"],
                    "C": ["small", "large", "large", "small",
                          "small", "large", "small", "small",
                          "large"],
                    "D": [1, 2, 2, 3, 3, 4, 5, 6, 7],
                    "E": [2, 4, 5, 5, 6, 6, 8, 9, 9]})
```

```python
df
```

```python
res1=pd.pivot_table(df, index='A', columns= 'C', values= 'D', aggfunc= 'sum')
```

```python
res1
```

```python
res2 = pd.pivot_table(df, index='A', columns= 'C', values= 'D', aggfunc= list)
res2   # aggfunc=list: Замість того, щоб рахувати суму чи середнє, Pandas просто збирає всі знайдені числа в список і кладе їх у відповідну клітинку
```

```python
res3 = pd.pivot_table(df, index='A', columns='C', values=['D'], aggfunc="sum")
res3
```

```python
display(res1.columns, res3.columns)
```

```python
# Як позбутися MultiIndex
res3.columns.droplevel(0)
```

```python
res3.columns = res3.columns.droplevel(0)
```

```python
res3
```

```python
display(res1.columns, res3.columns)  # Візуальна перевірка та порівняння структур стовпців обох таблиць
```
